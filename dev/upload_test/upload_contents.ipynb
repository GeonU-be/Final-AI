```python
# SNS 업로드 홍보글 자동 생성

import requests
import os
from dotenv import load_dotenv

# .env 파일 읽기
load_dotenv()

# 환경 변수에서 API 키 가져오기
API_KEY = os.getenv("UPSTAGE_API_KEY")

# 홍보글 생성
def generate_solar(product_name, keyword, platform, tone):
    prompt = f"""
SNS 홍보글 생성:

제품명: {product_name}
키워드: {keyword}
플랫폼: {platform}
톤: {tone}

조건:
- 해시태그 7개 이상
- 자연스럽게 작성
- 홍보글만 출력
"""

    headers = {
        "Authorization": f"Bearer {API_KEY}",
        "Content-Type": "application/json"
    }

    data = {
        "model": "solar-pro",
        "messages": [
            {"role": "user", "content": prompt}
        ]
    }

    res = requests.post(
        "https://api.upstage.ai/v1/chat/completions",
        headers=headers,
        json=data
    )

    return res.json()["choices"][0]["message"]["content"]

result = generate_solar(
    "극강보온 패딩",
    "보온, 방한, 패딩",
    "Instagram",
    "따뜻한 트렌디"
)

print(result)
```

```python
# 네이버 블로그용 홍보글 자동 생성 테스트 코드

import os
import requests
from dotenv import load_dotenv

load_dotenv()

API_KEY = os.getenv("UPSTAGE_API_KEY")

def generate_blog_post(product_name, keyword, tone="전문적이고 친근하며 사람들의 구매욕을 향상시키는 스타일"):

    prompt = f"""

당신은 네이버 블로그 홍보글 전문 카피라이터입니다.
다음 조건에 맞춰 **2000~3000자 수준의 매우 길고 상세한 네이버 블로그 최적화 홍보글**을 작성하세요

[제품 정보]
- 제품명 : {product_name}
- 키워드 : {keyword}

[작성 규칙]
1. 총 분량은 반드시 **2000자 이상**으로 작성할 것
2. 전체 구성: 서론 → 본문 소제목 5~6개 → 결론 → 해시태그
3. 각 소제목은 <h3> 태그 사용
4. 각 소제목 아래 최소 3~5 문단 작성
5. 설명, 후기 인용, 사용 상황 예시 포함
6. HTML 태그 적극 활용: <b>, <br>, <h3>, <ul>, <li>
7. 이미지 삽입 위치만 표시 ( 실제 업로드 x )
    -> (이미지: 제품 이미지 삽입 위치)
8. SEO 최적화 키워드는 자연스럽게 본문 전체에 분산
9. 말투는 {tone}
10. 마지막에 **해시태그 최소 20개 이상**

[출력 형식]
- 네이버 블로그에 그대로 업로드 가능한 HTML 포함 텍스트만 출력
"""

    headers = {
            "Authorization": f"Bearer {API_KEY}",
            "Content-Type": "application/json"
        }

    data = {
            "model": "solar-pro",
            "messages": [
                {"role": "user", "content": prompt}
         ]
    }

    response = requests.post(
            "https://api.upstage.ai/v1/chat/completions",
            headers=headers,
            json=data
    )

    res_body = response.json()

    if "choices" not in res_body:
        print("ERROR:", res_body)
        return "API 오류 발생"

    return res_body["choices"][0]["message"]["content"]


# 테스트 실행
result = generate_blog_post(
    product_name="겨울철 추위를 달래줄 핫팩",
    keyword="보온성, 핫팩, 프리미엄, 겨울 버티기"
)

print("\n=== 네이버 블로그용 홍보글 ===\n")
print(result)
```
