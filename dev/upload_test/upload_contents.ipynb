```python
import requests
import os
from dotenv import load_dotenv

# .env 파일 읽기
load_dotenv()

# 환경 변수에서 API 키 가져오기
API_KEY = os.getenv("UPSTAGE_API_KEY")

# 홍보글 생성
def generate_solar(product_name, keyword, platform, tone):
    prompt = f"""
SNS 홍보글 생성:

제품명: {product_name}
키워드: {keyword}
플랫폼: {platform}
톤: {tone}

조건:
- 해시태그 7개 이상
- 자연스럽게 작성
- 홍보글만 출력
"""

    headers = {
        "Authorization": f"Bearer {API_KEY}",
        "Content-Type": "application/json"
    }

    data = {
        "model": "solar-pro",
        "messages": [
            {"role": "user", "content": prompt}
        ]
    }

    res = requests.post(
        "https://api.upstage.ai/v1/chat/completions",
        headers=headers,
        json=data
    )

    return res.json()["choices"][0]["message"]["content"]

result = generate_solar(
    "극강보온 패딩",
    "보온, 방한, 패딩",
    "Instagram",
    "따뜻한 트렌디"
)

print(result)
```

```python
# FastAPI 없이 업로드 MOCK 로직

import os

# Twitter 업로드 MOCK 함수

def mock_upload_to_twitter(text: str, image_path: str = None):
    result = {
        "platform": "twitter",
        "status": "mock_success",
        "uploaded_text": text,
        "image" : None
    }

    if image_path:
        if os.path.exists(image_path):
            result["image"] = os.path.basename(image_path)
        else:
            result["image"] = "이미지 파일 없음"

    return result


# Instagram upload MOCK

def mock_upload_to_instagram(text: str, image_path: str = None):
    result = {
        "platform": "instagram",
        "status": "mock_success",
        "uploaded_text": text,
        "image" : None
    }

    if image_path:
        if os.path.exists(image_path):
            result["image"] = os.path.basename(image_path)
        else:
            result["image"] = "이미지 파일 없음"

    return result


# Start Test

twitter_result = mock_upload_to_twitter(
    text="twitter mock upload test",
    image_path=None
)

instagram_result = mock_upload_to_instagram(
    text="instagram mock upload test",
    image_path=None
)

print("Twitter Upload MOCK result\n", twitter_result)
print("Instagram Upload MOCK result\n", instagram_result)
```
