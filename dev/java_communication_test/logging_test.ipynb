```python
from dotenv import load_dotenv, find_dotenv
import os

# Load closest .env (works whether running from repo root or dev/)
load_dotenv(find_dotenv(), override=False)

url = os.getenv("JAVA_SERVER_ADDRESS")
```

```python
message = "테스트 메세지 입니다."
submessage = "테스트 중... 2"
logType = "INFO"
loggedProcess = "Start"
```

```python
import requests
from datetime import datetime

def send_log(
    message: str,
    submessage: str = "",
    logType: str = "INFO",
    loggedProcess: str = "Log",
    jobId: str = " "):
  try:
    return requests.post(
      url=f"{url}/api/log",
      json={
        "userId": "1",
        "logType": logType,
        "loggedProcess": loggedProcess,
        "loggedDate": datetime.now().isoformat(),
        "message": message,
        "submessage": submessage,
        "jobId": jobId
      }
    )
  except requests.exceptions.RequestException as e:
    print(f"로깅 실패!: ", e)
    raise
```

```python
res = send_log(message=message, submessage=submessage)
```

```python
print(res.request.method)
print(res.request.headers)
print("body: ")
print(res.request.body.decode(encoding="utf-8"))
```

```python
print(res.status_code)
print(res)
```
